```python
df3_clean = df3.copy()

df3_clean['STATE/UT'] = df3_clean['STATE/UT'].str.strip().str.upper()
df3_clean['DISTRICT'] = df3_clean['DISTRICT'].str.strip().str.upper()
```

```python
df3_clean['ANNUAL'].agg(['mean', 'median', 'min', 'max']).round(2)
```

```python
df3_clean[['STATE/UT', 'DISTRICT', 'ANNUAL']] \
.sort_values('ANNUAL', ascending=False).head(10)
```

```python
df3_clean[['STATE/UT', 'DISTRICT', 'ANNUAL']].sort_values('ANNUAL', ascending=True).head(10)
```

```python
df3_clean.groupby('STATE/UT')['ANNUAL'].mean().sort_values(ascending=False).head(10).round(2)
```

```python
df3_clean.groupby('STATE/UT')['JJAS'].mean().sort_values(ascending=False).head(10).round(2)
```
