```python
# Exploratory Data Analysis (EDA)
```

```python
# Question: Which month received the highest total rainfall?

df_clean.groupby("Month")["Daily Actual"].sum().sort_values(ascending=False)
```

```python
# Question: Which 10 districts received the highest total recorded rainfall?

df_clean.groupby("District")["Daily Actual"].sum().sort_values(ascending=False).head(10)
```

```python
# Question: Which 10 districts received the lowest total recorded rainfall?

df_clean.groupby("District")["Daily Actual"].sum().sort_values().head(10)
```

```python
# Question: Which states had the highest average daily rainfall?

df_clean.groupby("State")["Daily Actual"].mean().sort_values(ascending=False).head(10)
```

```python
# Question: How does actual rainfall compare with normal rainfall across each month?

df_clean.groupby("Month")[["Daily Actual", "Daily Normal"]].sum()
```

```python
# Question: Which districts had the highest average daily rainfall?

df_clean.groupby("District")["Daily Actual"].mean().sort_values(ascending=False).head(10)
```

```python
# Question: Which 10 dates recorded the highest daily rainfall?

df_clean.groupby("Date")["Daily Actual"].sum().sort_values(ascending=False).head(10)
```

```python
# Question: Which districts had the highest rainfall variability?

df_clean.groupby("District")["Daily Actual"].std().sort_values(ascending=False).head(10)
```

```python
# Question: Which districts had the highest number of dry days?

df_clean[df_clean["Daily Actual"] == 0].groupby("District").size().sort_values(ascending=False).head(10)
```

```python
# Question: Which 10 districts had the largest negative rainfall departure from normal?

df_clean.groupby("District")[["Daily Actual", "Daily Normal"]].sum().assign(
    Departure=lambda x: x["Daily Actual"] - x["Daily Normal"]
).sort_values("Departure").head(10)
```

```python
# Question: Which 10 districts had the largest positive rainfall departure from normal?

df_clean.groupby("District")[["Daily Actual", "Daily Normal"]].sum().assign(
    Departure=lambda x: x["Daily Actual"] - x["Daily Normal"]
).sort_values("Departure", ascending=False).head(10)
```

```python
# Question: Which month had the largest percentage departure from normal?

monthly_departure = df_clean.groupby("Month")[["Daily Actual", "Daily Normal"]].sum()

monthly_departure["Departure %"] = (
    (monthly_departure["Daily Actual"] - monthly_departure["Daily Normal"])
    / monthly_departure["Daily Normal"]
) * 100

monthly_departure.sort_values("Departure %")
```

```python
# Question: How many days of rainfall data are available for each district?

district_coverage = df_clean.groupby("District")["Date"].nunique().sort_values()

district_coverage.head(10)
```

```python
# Question: Which districts had the highest number of zero-rainfall days?

dry_days = (
    df_clean[df_clean["Daily Actual"] == 0]
    .groupby("District")
    .size()
    .sort_values(ascending=False)
)

dry_days.head(10)
```
